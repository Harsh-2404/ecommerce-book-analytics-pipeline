# Capstone Project - Phase 2: Data Cleaning & Preprocessing

**Dataset:** Books to Scrape  
**Source:** https://books.toscrape.com/  
**Input File:** raw_books_data.csv  
**Objective:** Clean, standardize, and transform the raw scraped dataset into a structured and analysis-ready dataset.

Import Libraries

```python
# Import required libraries

import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

Upload Raw CSV

```python
# Upload the raw dataset

# from google.colab import files

# uploaded = files.upload()
```

Load Raw Dataset

```python
# Load the raw dataset

df = pd.read_csv("raw_books_data.csv")

print("Raw dataset loaded successfully.")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

Display Dataset

```python
# Display the first five records

display(df.head())
```

Standardize Column Names

```python
# Standardize column names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Column names standardized successfully.")
print(df.columns.tolist())
```

Clean Text Columns

```python
# Clean and standardize text columns

df["title"] = df["title"].astype(str).str.strip()
df["stock_status"] = df["stock_status"].astype(str).str.strip().str.title()
df["product_url"] = df["product_url"].astype(str).str.strip()

print("Text columns cleaned successfully.")
```

Ensure Correct Numeric Data Types

```python
# Convert numerical columns to appropriate data types

df["price_gbp"] = pd.to_numeric(df["price_gbp"], errors="coerce")
df["rating_stars"] = pd.to_numeric(df["rating_stars"], errors="coerce")

print("Numeric data types standardized successfully.")
```

Handle Missing Values

```python
# Handle missing values

missing_values = df.isnull().sum()      # sum of all data in each Column

print("Missing values by column:")
print(missing_values)

if missing_values.sum() == 0:         # Total Sum of All Columns
    print("\nNo missing values found. No imputation was required.")
else:
    print("\nMissing values detected and will be handled appropriately.")
```

Handle Duplicate Records

```python
# Remove duplicate records

duplicate_count = df.duplicated().sum()

print(f"Duplicate records found: {duplicate_count}")

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicate records removed successfully.")
else:
    print("No duplicate records required removal.")
```

Create Useful Feature: Rating Percentage

```python
# Create rating percentage feature

df["rating_percentage"] = (df["rating_stars"] / 5) * 100

print("Rating percentage feature created successfully.")
```

Create Price Band

```python
# Create price band feature

def classify_price(price):
    if price < 20:
        return "Low"
    elif price < 40:
        return "Medium"
    elif price < 60:
        return "High"
    else:
        return "Premium"

df["price_band"] = df["price_gbp"].apply(classify_price)

print("Price band feature created successfully.")
```

Create Stock Availability Flag

```python
# Create stock availability flag

df["in_stock_flag"] = df["stock_status"].str.lower().eq("in stock").astype(int)

print("Stock availability flag created successfully.")
```

Final Clean Dataset Preview

```python
# Display the cleaned dataset

display(df.head(10))
```

Save Clean Dataset

```python
# Save the cleaned dataset

clean_file = "clean_books_data.csv"

df.to_csv(clean_file, index=False)

print(f"Cleaned dataset saved successfully as: {clean_file}")
```
